```python
import json

with open ('data.json') as f:
    data = json.load(f)
```

PRODUCTO CON MÁS INGRESOS

```python
ingresos = {}

for venta in data["ventas"]:
    prod = venta["producto"]
    ingresos[prod] = ingresos.get(prod, 0) + venta["cantidad"] * venta["precio_unitario"]

mas_ingresos = max(ingresos, key=ingresos.get)
print(mas_ingresos, ingresos[mas_ingresos])
```

FILTRAR VENTAS DE ANA

```python
vendedor = "Ana"

ventas_ana = [venta for venta in data["ventas"] if venta["vendedor"] == vendedor]

print(ventas_ana)
```

CALCULAR EL TICKET MEDIO

```python
total = sum(v["cantidad"] * v["precio_unitario"] for v in data["ventas"])
num_ventas = len(data["ventas"])

ticket_medio = total / num_ventas
print(f"Ticket medio: {ticket_medio:.2f} €")
```
